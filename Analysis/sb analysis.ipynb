```python
import h5py
import numpy as np
import plotly.express as px
```

```python
f = r"C:\DATA\2025_10_01_tpx\testing\servalcode_test1_2025-10-01_11-49\position_000000_gated_mean.tpx3.cv4"

with h5py.File(f, "r") as f:
    x, y, t = f['x'][()], f['y'][()], f['t'][()]
    pulse = f['pulses'][()]

```

```python
px.density_heatmap(
        x=x,
        y=y,
        nbinsx=256,
        nbinsy=256,
        width=800,
        height=800,
).show()
hist, xe = np.histogram(t, range=[3000, 5000], bins=4000)
peak = xe[np.argmax(hist)]
print(peak)
edge = xe[np.argmin(np.abs(hist - np.max(hist) / 2))]
hwhm = peak - edge
tf = t[(t < 5000) & (t > 3000)]

px.histogram(x=tf, range_x=[3000, 5000], nbins=1000, log_y=False).add_vline(x=peak - hwhm).add_vline(peak + hwhm).show()
print(2 * hwhm)
```

```python
rf = np.sqrt((x - 138) ** 2 + (y - 133) ** 2)[t < 8000]
# rf=x[t<10000]
tf = t[t < 8000]
rf = rf[tf > 4000]
tf = tf[tf > 4000]

px.density_heatmap(
        y=rf,
        x=tf / 1000,
        nbinsx=1000,
        nbinsy=256,
        range_x=[5, 8],
        width=800,
        height=800,
).show()
```
