```python
import h5py
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
f = r"C:\DATA\StonyBrookCollab\2025_10_01_tpx\CHD\servalcode_test1_2025-10-01_11-49.cv4"

with h5py.File(f, "r") as f:
    x, y, t = f['x'][()], f['y'][()], f['t'][()]
    cluster_corr = f['cluster_corr'][()]
    pulse = f['pulses'][()]
    for k in f.keys():
        print(k)
        print(f[k].shape)

serval_data = pd.DataFrame({"#x": x, "#y": y, "#tof": t / 1e9})
serval_data["#r"] = np.sqrt((x - 138) ** 2 + (y - 133) ** 2)
```

```python
px.density_heatmap(
        x=x,
        y=y,
        nbinsx=256,
        nbinsy=256,
        width=800,
        height=800,
).show()
hist, xe = np.histogram(t, range=[3000, 5000], bins=4000)
peak = xe[np.argmax(hist)]
print(peak)
edge = xe[np.argmin(np.abs(hist - np.max(hist) / 2))]
hwhm = peak - edge
tf = t[(t < 5000) & (t > 3000)]

px.histogram(x=tf, range_x=[3000, 5000], nbins=1000, log_y=False).add_vline(x=peak - hwhm).add_vline(peak + hwhm).show()
print(2 * hwhm)
```

```python
rf = np.sqrt((x - 138) ** 2 + (y - 133) ** 2)[t < 4000]
tf = t[t < 4000]
rf = rf[tf > 3500]
tf = tf[tf > 3500]

px.density_heatmap(
        y=rf,
        x=tf / 1000,
        nbinsx=1000,
        nbinsy=256,
        range_x=[3.5, 4],
        width=800,
        height=800,
).show()
```

```python
pymepix_file = r"C:\DATA\StonyBrookCollab\2025_10_01_tpx\CHD\pymepix_test1_2025-10-01_11-30_CENT.csv"

ppx_data = pd.read_csv(pymepix_file)
ppx_data['#r'] = np.sqrt((ppx_data['#x'] - 138) ** 2 + (ppx_data["#y"] - 133) ** 2)
```

```python
ppx_idx = (3000 < ppx_data["#tof"] * 1e9) & (ppx_data["#tof"] * 1e9 < 5000)
# ppx_idx=(0000<ppx_data["#tof"]*1e9) & (ppx_data["#tof"]*1e9<10000)
px.histogram(ppx_data[ppx_idx], x="#tof", log_y=True).show()

serval_idx = (3000 < serval_data["#tof"] * 1e9) & (serval_data["#tof"] * 1e9 < 5000)
# serval_idx=(000<serval_data["#tof"]*1e9) & (serval_data["#tof"]*1e9<10000)
px.histogram(serval_data[serval_idx], x="#tof", log_y=True).show()

px.density_heatmap(
        ppx_data[ppx_idx],
        x="#tof",
        y="#r",
        nbinsx=512,
        nbinsy=512,
).show()

px.density_heatmap(
        serval_data[serval_idx],
        x="#tof",
        y="#r",
        nbinsx=512,
        nbinsy=512,
).show()
```

```python
px.density_heatmap(serval_data[serval_idx],
                   x="#x", y="#y", z="#tof",
                   nbinsx=256, nbinsy=256,
                   histfunc="avg", range_color=[3.8e-6, 4.2e-6],
                   color_continuous_scale=px.colors.sequential.Inferno,
                   ).show()

px.density_heatmap(ppx_data[ppx_idx],
                   x="#x", y="#y", z="#tof",
                   nbinsx=256, nbinsy=256,
                   histfunc="ag", range_color=[4.25e-6, 4.35e-6],
                   color_continuous_scale=px.colors.sequential.Inferno,
                   ).show()
```
