```python

import os

import pandas as pd
import plotly.express as px

dir = r"J:\ctgroup\Edward\DATA\VMI\20241120\calibrated"

files = [f for f in os.listdir(dir) if f.endswith('.h5')]
files = ['ch4_p_6W_calibrated.h5']
targets = [f.split('_')[0] for f in files]
powers = [f.split('_')[2] for f in files]
```

```python
for file, target, power in zip(files, targets, powers):
    print(file)
    data = pd.read_hdf(os.path.join(dir, file))
    data = data[(data['m/q'] > 0) & (data['m/q'] < 60)]
    px.histogram(data['m/q'], nbins=1000, title=f'{target} at {power}',
                 labels={'value': 'm/q (amu/e)', 'count': 'Counts'}).show()
```
