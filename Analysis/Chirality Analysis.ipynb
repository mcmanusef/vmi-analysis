```python
import os

import numpy as np
import pandas as pd
import plotly.express as px

dir = r"J:\ctgroup\Edward\DATA\VMI\20250303\Propylene Oxide 2W_calibrated"
```

```python
all_data = pd.DataFrame()
for file in os.listdir(dir):
    if not file.endswith('.h5'):
        continue
    angle = float(file.split('_')[0])
    data = pd.read_hdf(os.path.join(dir, file))
    data['angle'] = angle
    data['pr'] = np.sqrt(data['px'] ** 2 + data['py'] ** 2 + data['pz'] ** 2)
    all_data = pd.concat([all_data, data], ignore_index=True)
```

```python
filtered_data = all_data[(all_data['pr'] > 0) & (all_data['pr'] < 0.6)]
grouped = filtered_data.groupby('angle')
mean_py = grouped['py'].mean()
px.scatter(mean_py, title='Mean py vs Angle', labels={'index': 'Angle (degrees)', 'py': 'Mean py (a.u.)'}).show()
mean_px = grouped['px'].mean()
px.scatter(mean_px, title='Mean px vs Angle', labels={'index': 'Angle (degrees)', 'px': 'Mean px (a.u.)'}).show()
mean_pz = grouped['pz'].mean()
px.scatter(mean_pz, title='Mean pz vs Angle', labels={'index': 'Angle (degrees)', 'pz': 'Mean pz (a.u.)'}).show()
```
