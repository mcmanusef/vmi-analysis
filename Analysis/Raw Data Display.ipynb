```python
import numpy as np
import pandas as pd
import plotly.express as px
import h5py

file=r"J:\ctgroup\Edward\DATA\VMI\20260430\PO_E_1.cv4"
data = dict()
with h5py.File(file, 'r') as f:
    for key in f.keys():
        print(key)
        data[key] = f[key][()]
```

```python
cluster_df= pd.DataFrame({
    'x':data['x'],
    'y':data['y'],
    't':data['t'],
    'corr':data['cluster_corr']
})
etof_df = pd.DataFrame({
    'etof':data['t_etof']+np.random.random_sample(len(data['t_etof']))*0.26,
    'corr':data['etof_corr'],
})
itof_corr_df = pd.DataFrame({
    'itof':data['t_tof'],
    'corr':data['tof_corr'],
})
```

```python
cluster_df=cluster_df[cluster_df['t'].between(0,1000)]
px.histogram(cluster_df, x='t', nbins=20000).show()
px.density_heatmap(
        cluster_df,
        x='x', y='y',
        nbinsx=256, nbinsy=256,
        color_continuous_scale='Inferno',
        height=1000, width=1000
                   ).show()
```

```python
etof_df=etof_df[etof_df['etof'].between(450,500)]
px.histogram(etof_df, x='etof', nbins=20000).show()
```

```python
combined_df = pd.merge_ordered(cluster_df, etof_df, on='corr', how='inner')
px.density_heatmap(
        combined_df, x='x', y='y',
        nbinsx=256, nbinsy=256,
        color_continuous_scale='Inferno',
        height=1000, width=1000
).show()

px.histogram(
        combined_df[combined_df['x'].between(125,135) & combined_df['y'].between(125,135)],
        x='etof', nbins=1000
).show()

px.density_heatmap(
        combined_df,
        x='x', y='etof',
        nbinsx=256, nbinsy=256,
        color_continuous_scale='Inferno',
        height=1000, width=1000
).show()
```

```python
import numpy as np
import pandas as pd
import plotly.express as px
import h5py

file=r"J:\ctgroup\Edward\DATA\VMI\20260430\PO_E_1.cv4"
data = dict()
with h5py.File(file, 'r') as f:
    for key in f.keys():
        print(key)
        data[key] = f[key][()]
```

```python
pixel_df= pd.DataFrame({
    'x':data['x'],
    'y':data['y'],
    't':data['t'],
    # 'tot':data['tot'],
})
pixel_df=pixel_df[pixel_df['t'].between(0,1000)]
pixel_df=pixel_df[(pixel_df['x']<200 ) | (pixel_df['y']<200)]
px.density_heatmap(
        pixel_df.sample(frac=0.5), x='x', y='y',
        nbinsx=256, nbinsy=256,
        color_continuous_scale='Inferno',
        height=800, width=900,
        histfunc="count"
).show()
```
