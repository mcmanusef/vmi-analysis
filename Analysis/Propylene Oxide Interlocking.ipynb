```python
import pandas as pd
f=r"J:\ctgroup\Edward\DATA\VMI\20250307\Intensity Scan\7_calibrated.h5"
df=pd.read_hdf(f,"data")
df=df[(df['px']**2+df['py']**2+df['pz']**2).between(0,0.41**2)]
# df=df[df['m/q'].between(55,65)]
```

```python
import plotly.express as px
import numpy as np
import skimage

df_filt=df[
    df['py'].between(-0.1,0.1)
    #&~df['m/q'].between(55,65)
]

hist2d, xedges, yedges = np.histogram2d(
    df_filt['px'], df_filt['pz'],
    bins=256,
    density=True,
)
x_vec=(xedges[:-1]+xedges[1:])/2
y_vec=(yedges[:-1]+yedges[1:])/2


im=hist2d.T
im= skimage.filters.median(im, footprint=skimage.morphology.disk(3))
im=skimage.filters.difference_of_gaussians(
    im,
    low_sigma=1,
    high_sigma=30,
)
# im= skimage.filters.gaussian(im, sigma=1)

fig=px.imshow(
    im,
    x=x_vec,
    y=y_vec,
    origin='lower',
    labels={'x':'px (a.u.)', 'y':'pz (a.u.)'},
    color_continuous_scale=px.colors.sequential.Inferno,
    zmin=0,
)

fig.update_layout(coloraxis_colorbar=dict(
    title="Signal\n(arb. units)",
    ticks="outside",
    title_side="right",
),
template='simple_white+presentation',
        height=400,
        width=500,
)
fig.show()
fig.write_image("Propylene_Oxide_Interlocking.png", scale=15)
```
